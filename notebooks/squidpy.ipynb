```python
import scanpy as sc
import spatialdata as sd
import squidpy as sq
import spatialdata_plot  # noqa: F401
from spatialdata_io import merscope
```

Use [spatialdata-io]() to convert public datasets to spatialdata formats with:

`python -m spatialdata_io ..`

Requires the following files:

- 'cell_by_gene.csv': Counts file.
- 'detected_transcripts.csv': Transcript file.
- 'cell_metadata.csv': Per-cell metadata file.
- 'cell_boundaries.parquet': Cell polygon boundaries.
- mosaic_**_z*.tif images inside the 'images' directory.

```python
from spatialdata._core.spatialdata import SpatialData
raw_data_path = 'path/to/raw/data'
zarr_data_pth = 'path_write'

sdata = merscope(
    path = raw_data_path,
    transcripts = True,
    cells_boundaries = True,
    cells_table = True,
    mosaic_images = True,
    backend = None # By default, uses "rioxarray" if and only if the rioxarray library is installed (reduces RAM usage).
)

# write the data to disk
sdata.write(zarr_data_pth)

# read and print the Zarr data
sdata: SpatialData = sd.read_zarr(zarr_data_pth)
print(sdata)
```
